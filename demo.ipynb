#### for loop executes a set of operations on each elements from one collection

```python
distance_list = [1000, 1500, 1700]  #in metres (m)

# to convert m to km, we need to divide each elements by 1000

distance_km_list = distance_list / 1000
```

```python
first_distance_in_km = distance_list[0] / 1000
print(first_distance_in_km)
second_distance_in_km = distance_list[1] / 1000
print(second_distance_in_km)
```

```python
# writing for loops

for distance in distance_list:
    distance_km = distance / 1000
    print(distance_km)
```

```python
# change the negative to positive value 

numbers = [1.0, -2.3, 3.5, -1.9]

positive_list = []
for number in numbers:
    #print(number)
    if number < 0:
        new_num = number * (-1)
    else:
        new_num = number
    positive_list.append(new_num)

print(positive_list)
    
        
```

```python
# change the negative to positive value and a cut off

numbers = [1.0, -2.3, 3.5, -1.9]

positive_list = []
for number in numbers:
    #print(number)
    if number < 0:
        new_num_pre = number * (-1)
        if new_num_pre > 2:
            new_num = 2
        
    elif number > 0:
        new_num = number
    elif number > 2:
        new_num = 2
    positive_list.append(new_num)

print(positive_list)
    
```

## Challange: Collatz conjecture


```python
## Challange: Collatz conjecture
i = 122
while i!=1:
    if (i%2 == 0):
        i = i/2
    else:
        i = 3 * i + 1
    print(int(i))
```

```python
number = 29
step = 200
for i in range(step):
    if number % 2 == 0:
        number = number/2
    else :
        number = 3 * number + 1
    print('iteration = ',i, 'number = ', int(number))
    if number == 1:
        break
```
